```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Helllo")
plt.xlabel("Chacha")
plt.ylabel("Chachi")
```

```python
plt.subplot(2,2,1)
plt.plot(x,x)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,y)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,16)
plt.subplot(6,6,4)
plt.subplot(6,6,1) 
plt.subplot(6,6,9) 
plt.subplot(6,6,12)
```
